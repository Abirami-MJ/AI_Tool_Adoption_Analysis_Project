```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display all columns
pd.set_option('display.max_columns', None)

df = pd.read_csv("AI_Adoption_Research_Grade_350.csv")
```

```python
# Correlation analysis

numerical_cols = [
    'Employee_Count',
    'AI_Adoption_Score',
    'Monthly_AI_Users',
    'Productivity_Gain_Percent',
    'Satisfaction_Score',
    'AI_Investment_USD',
    'Annual_Cost_Savings_USD',
    'ROI_Percent',
    'Training_Hours_Per_Employee',
    'Automation_Level_Percent'
]

correlation_matrix = df[numerical_cols].corr()

print(correlation_matrix)
```

```python
# correlation heatmap

import matplotlib.pyplot as plt
import seaborn as sns

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Optional for better-looking plots
sns.set_style("whitegrid")

plt.figure(figsize=(12,8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title("Correlation Heatmap of AI Adoption Variables")

plt.show()
```

```python
#Adoption VS productivity

plt.figure(figsize=(8,5))

sns.scatterplot(
    x='AI_Adoption_Score',
    y='Productivity_Gain_Percent',
    data=df
)

plt.title("AI Adoption Score vs Productivity Gain")

plt.show()
```

```python
#investment VS saving
plt.figure(figsize=(8,5))

sns.scatterplot(
    x='AI_Investment_USD',
    y='Annual_Cost_Savings_USD',
    data=df
)

plt.title("AI Investment vs Cost Savings")

plt.show()
```

```python
#employee count VS adoption
plt.figure(figsize=(8,5))

sns.scatterplot(
    x='Employee_Count',
    y='AI_Adoption_Score',
    data=df
)

plt.title("Employee Count vs AI Adoption Score")

plt.show()
```

```python
# satisfaction VS productivity

plt.figure(figsize=(8,5))

sns.scatterplot(
    x='Satisfaction_Score',
    y='Productivity_Gain_Percent',
    data=df
)

plt.title("Satisfaction Score vs Productivity Gain")

plt.show()
```
