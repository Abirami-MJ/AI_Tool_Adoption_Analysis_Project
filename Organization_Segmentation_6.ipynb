```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display all columns
pd.set_option('display.max_columns', None)

df = pd.read_csv("AI_Adoption_Research_Grade_350.csv")
```

```python
#Segmentation 1: AI Leaders vs Slow Adopters

def adoption_category(score):
    if score >= 85:
        return "AI Leader"
    elif score >= 65:
        return "Early Adopter"
    else:
        return "Slow Adopter"

df['Adoption_Category'] = df['AI_Adoption_Score'].apply(adoption_category)

df[['AI_Adoption_Score','Adoption_Category']].head()
```

```python
#Segmentation 2: High ROI Organizations

def roi_category(roi):
    if roi >= 150:
        return "High ROI"
    elif roi >= 75:
        return "Medium ROI"
    else:
        return "Low ROI"

df['ROI_Category'] = df['ROI_Percent'].apply(roi_category)
df['ROI_Category'].head()
```

```python
#Segmentation 3: Satisfaction Groups

def satisfaction_group(score):
    if score >= 8:
        return "Highly Satisfied"
    elif score >= 5:
        return "Moderately Satisfied"
    else:
        return "Low Satisfaction"

df['Satisfaction_Group'] = df['Satisfaction_Score'].apply(satisfaction_group)
df['Satisfaction_Group'].head()
```

```python
#Segmentation 4: Automation Maturity

def automation_group(level):
    if level >= 80:
        return "Highly Automated"
    elif level >= 50:
        return "Moderately Automated"
    else:
        return "Low Automation"

df['Automation_Group'] = df['Automation_Level_Percent'].apply(automation_group)
df['Automation_Group'].head()
```
