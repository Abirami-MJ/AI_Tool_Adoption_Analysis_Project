```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display all columns
pd.set_option('display.max_columns', None)

df = pd.read_csv("AI_Adoption_Research_Grade_350.csv")

#INDUSTRY VS AI ADOPTION
industry_adoption = df.groupby('Industry')['AI_Adoption_Score'].mean().sort_values(ascending=False)

plt.figure(figsize=(10,5))

sns.barplot(
    x=industry_adoption.index,
    y=industry_adoption.values
)

plt.title("Average AI Adoption Score by Industry")
plt.xlabel("Industry")
plt.ylabel("Average Adoption Score")
plt.xticks(rotation=45)

plt.show()
```

```python
#COMPANY SIZE VS ADOPTION

plt.figure(figsize=(8,5))

sns.boxplot(
    x='Company_Size',
    y='AI_Adoption_Score',
    data=df
)
plt.title("Company Size vs AI Adoption Score")

plt.show()
```

```python
company_adoption = df.groupby('Company_Size')['AI_Adoption_Score'].mean()

plt.figure(figsize=(6,5))

sns.barplot(
    x=company_adoption.index,
    y=company_adoption.values
)

plt.title("Average Adoption Score by Company Size")

plt.show()
```

```python
#AI tools vs satisfaction

tool_satisfaction = (
    df.groupby('AI_Tool')['Satisfaction_Score']
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10,5))

sns.barplot(
    x=tool_satisfaction.index,
    y=tool_satisfaction.values
)

plt.title("Average Satisfaction Score by AI Tool")
plt.xticks(rotation=45)

plt.show()
```

```python
#adoption year vs growth

year_growth = df.groupby('Adoption_Year').size()

plt.figure(figsize=(10,5))

sns.lineplot(
    x=year_growth.index,
    y=year_growth.values,
    marker='o'
)

plt.title("AI Adoption Growth Over Years")
plt.xlabel("Adoption Year")
plt.ylabel("Number of Organizations")

plt.show()
```

```python
#Employee count vs Ai usage

plt.figure(figsize=(10,6))

sns.scatterplot(
    x='Employee_Count',
    y='Monthly_AI_Users',
    data=df
)

plt.title("Employee Count vs Monthly AI Users")

plt.show()
```

```python
# Investment vs Saving

plt.figure(figsize=(10,6))

sns.scatterplot(
    x='AI_Investment_USD',
    y='Annual_Cost_Savings_USD',
    data=df
)

plt.title("AI Investment vs Cost Savings")

plt.show()
```
